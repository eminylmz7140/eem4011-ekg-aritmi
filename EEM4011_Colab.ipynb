```python
from urllib.request import urlretrieve
from pathlib import Path
import zipfile, os
url = 'https://raw.githubusercontent.com/eminylmz7140/eem4011-ekg-aritmi/main/EEM4011_teslim_paketi.zip'
archive = Path('/content/EEM4011_teslim_paketi.zip')
urlretrieve(url, archive)
with zipfile.ZipFile(archive) as z:
    assert z.testzip() is None, 'ZIP bütünlük kontrolü başarısız'
    z.extractall('/content/eem4011')
os.chdir('/content/eem4011/biomed_project')
print('Çalışma klasörü:', Path.cwd())
```
